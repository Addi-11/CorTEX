````python
# Run TxAgent on each row to get tool calls with arguments AND results
# Capture stdout to extract tool call results that are only printed
import json
import sys
import io
import re

def parse_tool_calls_from_output(output_text):
    """Parse tool calls and results from TxAgent's printed output."""
    # Remove ANSI color codes
    ansi_escape = re.compile(r'\x1b\[[0-9;]*m')
    clean_text = ansi_escape.sub('', output_text)
    
    tool_calls_with_results = []
    
    # Split by "Tool Call:" to get individual tool call sections
    # Pattern: Tool Call: {dict} ... Tool Call Result: {result}
    sections = clean_text.split('Tool Call:')
    
    for section in sections[1:]:  # Skip first empty section
        try:
            # Find the tool call dict (ends at first newline or "Tool Call Result:")
            # The dict format is: {'name': '...', 'arguments': {...}}
            
            # Extract tool name and arguments using regex
            name_match = re.search(r"'name':\s*'([^']+)'", section)
            args_match = re.search(r"'arguments':\s*(\{[^}]+\})", section)
            
            if not name_match:
                continue
                
            tool_name = name_match.group(1)
            
            # Skip Finish and Tool_RAG
            if tool_name in ['Finish', 'Tool_RAG']:
                continue
            
            # Parse arguments
            arguments = None
            if args_match:
                try:
                    # Convert single quotes to double quotes for JSON
                    args_str = args_match.group(1).replace("'", '"')
                    arguments = json.loads(args_str)
                except:
                    arguments = args_match.group(1)
            
            # Find the result - it comes after "Tool Call Result:"
            result_match = re.search(r'Tool Call Result:\s*(.+?)(?=Tool Call:|Summarized Tool Result:|$)', section, re.DOTALL)
            
            result = None
            if result_match:
                result_str = result_match.group(1).strip()
                # Truncate very long results but try to parse JSON first
                if result_str.startswith('{') or result_str.startswith('['):
                    try:
                        # Try to parse as JSON
                        result = json.loads(result_str.split('\n')[0])  # Just first line if multiline
                    except:
                        # Truncate long results
                        result = result_str[:1000] + '...' if len(result_str) > 1000 else result_str
                else:
                    result = result_str[:1000] + '...' if len(result_str) > 1000 else result_str
            
            tool_calls_with_results.append({
                'name': tool_name,
                'arguments': arguments,
                'result': result
            })
            
        except Exception as e:
            continue
    
    return tool_calls_with_results

results = []

for idx, row in tqdm(df.iterrows(), total=len(df), desc="Running TxAgent inference"):
    query = row['input']
    
    try:
        # Capture stdout to get tool call results
        captured_output = io.StringIO()
        sys.stdout = captured_output
        
        # Run TxAgent to get tool calls
        response = agent.run_multistep_agent(
            query,
            temperature=0.3,
            max_new_tokens=1024,
            max_token=90240,
            call_agent=multiagent,
            max_round=max_round
        )
        
        # Restore stdout
        sys.stdout = sys.__stdout__
        output_text = captured_output.getvalue()
        
        # Parse tool calls and results from captured output
        tool_calls_with_results = parse_tool_calls_from_output(output_text)
        
        results.append({
            'index': idx,
            'instruction': row.get('instruction', ''),
            'input': query,
            'output': row['output'],
            'tool_calls': tool_calls_with_results,
            'final_response': response
        })
        
        # Print progress
        print(f"\n--- Sample {idx} ---")
        print(f"Query: {query[:100]}...")
        print(f"Tool Calls Captured: {len(tool_calls_with_results)}")
        for tc in tool_calls_with_results[:3]:
            print(f"  - {tc.get('name')}: {str(tc.get('arguments'))[:60]}...")
            if tc.get('result'):
                result_preview = str(tc.get('result'))[:80]
                print(f"    Result: {result_preview}...")
        
    except Exception as e:
        sys.stdout = sys.__stdout__  # Restore stdout on error
        results.append({
            'index': idx,
            'instruction': row.get('instruction', ''),
            'input': query,
            'output': row['output'],
            'tool_calls': None,
            'final_response': None,
            'error': str(e)
        })
        print(f"Error on sample {idx}: {e}")

results_df = pd.DataFrame(results)
print(f"\nCompleted inference on {len(results_df)} samples")
print(f"Total tool calls captured: {sum(len(tc) if tc else 0 for tc in results_df['tool_calls'])}")
results_df.head()
````

````python
# Create datasets for 2-model pipeline from TxAgent outputs
# Using SIMPLIFIED output format for better trainability
import json
import os
from collections import defaultdict

def create_training_datasets_v2(results_df, output_dir="datasets/finetuning"):
    """
    Create two datasets from TxAgent inference results with SIMPLIFIED formats:
    
    Model 1 (Tool Selection): Simpler text-based output format
    Model 2 (Reasoning): Question + tool results -> answer
    """
    os.makedirs(output_dir, exist_ok=True)
    
    model1_data = []  # Tool selection dataset
    model2_data = []  # Reasoning dataset
    
    for idx, row in results_df.iterrows():
        if not row.get('tool_calls') or len(row['tool_calls']) == 0:
            continue
            
        question = row['input']
        ground_truth = row['output']
        tool_calls = row['tool_calls']
        final_response = row.get('final_response', '')
        
        # ========== Model 1: Tool Selection Dataset (SIMPLIFIED) ==========
        # Group tool calls by tool name to show pattern
        tool_groups = defaultdict(list)
        for tc in tool_calls:
            name = tc.get('name', '')
            if name and name not in ['Finish', 'Tool_RAG']:
                args = tc.get('arguments', {})
                # Extract the key argument (usually drug_name)
                if isinstance(args, dict):
                    drug = args.get('drug_name', args.get('name', ''))
                    if drug:
                        tool_groups[name].append(drug)
                    else:
                        tool_groups[name].append(str(args))
                else:
                    tool_groups[name].append(str(args))
        
        if tool_groups:
            # Create simplified text output format
            # Format: TOOL_NAME: arg1, arg2, arg3
            output_lines = []
            for tool_name, args_list in tool_groups.items():
                unique_args = list(dict.fromkeys(args_list))  # Remove duplicates, preserve order
                output_lines.append(f"{tool_name}: {', '.join(unique_args)}")
            
            simplified_output = "\n".join(output_lines)
            
            model1_sample = {
                'instruction': "Identify the biomedical database tools and drug names needed to answer this medical question. Output format: TOOL_NAME: drug1, drug2, ...",
                'input': question,
                'output': simplified_output,
                'ground_truth_answer': ground_truth
            }
            model1_data.append(model1_sample)
        
        # ========== Model 2: Reasoning Dataset ==========
        # Input: Question + Tool results (formatted as readable text)
        # Output: Final answer with reasoning
        tool_results_text = []
        for tc in tool_calls:
            name = tc.get('name', '')
            if name and name not in ['Finish', 'Tool_RAG']:
                args = tc.get('arguments', {})
                result = tc.get('result', '')
                
                # Format arguments nicely
                if isinstance(args, dict):
                    args_str = ", ".join(f"{k}={v}" for k, v in args.items())
                else:
                    args_str = str(args)
                
                # Truncate very long results
                if isinstance(result, dict):
                    result = json.dumps(result)
                result_str = str(result)[:1500] if len(str(result)) > 1500 else str(result)
                
                tool_results_text.append(f"[{name}({args_str})]\n{result_str}")
        
        if tool_results_text and final_response:
            # Create readable input combining question and tool results
            model2_input = f"""QUESTION:
{question}

TOOL RESULTS:
{chr(10).join(tool_results_text)}"""
            
            model2_sample = {
                'instruction': "Based on the medical question and tool results, select the best answer option and explain your reasoning.",
                'input': model2_input,
                'output': final_response,
                'ground_truth_answer': ground_truth
            }
            model2_data.append(model2_sample)
    
    # Save datasets
    model1_path = os.path.join(output_dir, 'model1_tool_selection.json')
    model2_path = os.path.join(output_dir, 'model2_reasoning.json')
    
    with open(model1_path, 'w') as f:
        json.dump(model1_data, f, indent=2)
    
    with open(model2_path, 'w') as f:
        json.dump(model2_data, f, indent=2)
    
    print(f"✅ Model 1 (Tool Selection) dataset: {len(model1_data)} samples -> {model1_path}")
    print(f"✅ Model 2 (Reasoning) dataset: {len(model2_data)} samples -> {model2_path}")
    
    return model1_data, model2_data

# Generate datasets with simplified format
model1_data, model2_data = create_training_datasets_v2(results_df)

# Preview Model 1 sample
print("\n" + "="*80)
print("MODEL 1 SAMPLE - SIMPLIFIED FORMAT (Tool Selection)")
print("="*80)
if model1_data:
    sample = model1_data[0]
    print(f"INSTRUCTION: {sample['instruction']}")
    print(f"\nINPUT:\n{sample['input'][:300]}...")
    print(f"\nOUTPUT (what model learns to generate):\n{sample['output']}")
    print(f"\nGROUND TRUTH: {sample['ground_truth_answer']}")

# Preview Model 2 sample
print("\n" + "="*80)
print("MODEL 2 SAMPLE (Reasoning)")  
print("="*80)
if model2_data:
    sample = model2_data[0]
    print(f"INSTRUCTION: {sample['instruction']}")
    print(f"\nINPUT (first 500 chars):\n{sample['input'][:500]}...")
    print(f"\nOUTPUT (first 300 chars):\n{sample['output'][:300]}...")
````

## Dataset Format Summary

### Model 1: Tool Selection (LLaVA-Med Fine-tuning) - SIMPLIFIED FORMAT
**Purpose:** Learn which tools to call for a given medical question

**Why Simplified?** 
- Easier for model to learn consistent pattern
- Variable number of tools handled naturally
- No complex JSON escaping issues

| Field | Example |
|-------|---------|
| `instruction` | "Identify the biomedical database tools and drug names needed..." |
| `input` | Medical question with options |
| `output` | `FDA_get_pregnancy_effects_info_by_drug_name: Ampicillin, Nitrofurantoin` |

### Model 2: Reasoning Model
**Purpose:** Given question + tool results, reason to the correct answer

| Field | Description |
|-------|-------------|
| `instruction` | System prompt for reasoning task |
| `input` | Plain text: Question + formatted tool results |
| `output` | Final reasoning and answer selection |

### Pipeline Flow:
```
Question → Model 1 → "TOOL: drug1, drug2" → Parse → Execute APIs → Tool Results → Model 2 → Answer
```

### Advantages of Simplified Format:
1. **Consistent structure** - Always `TOOL_NAME: arg1, arg2, arg3`
2. **Easy to parse** - Simple string split on `:` and `,`
3. **Handles variable counts** - Model can output 1 or 20 drugs naturally
4. **No JSON errors** - Plain text is more robust than structured JSON
5. **Learnable pattern** - Model sees consistent format across all examples
